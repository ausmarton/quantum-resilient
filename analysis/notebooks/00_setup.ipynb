```python
# Check required libraries
import importlib

required_packages = [
    "pandas", "numpy", "scipy", "matplotlib", "seaborn",
    "tqdm", "pyarrow", "fastjsonschema", "rich",
]

optional_packages = [
    "google.cloud.storage", "google.cloud.bigquery",
    "boto3", "sklearn", "statsmodels",
]

print("Required packages:")
for pkg in required_packages:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, "__version__", "unknown")
        print(f"  ✓ {pkg}: {version}")
    except ImportError:
        print(f"  ✗ {pkg}: NOT INSTALLED")

print("\nOptional packages:")
for pkg in optional_packages:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, "__version__", "unknown")
        print(f"  ✓ {pkg}: {version}")
    except ImportError:
        print(f"  ○ {pkg}: not installed (optional)")
```

```python
# Setup directories
from pathlib import Path

ANALYSIS_ROOT = Path("..")
DATA_DIR = ANALYSIS_ROOT / "data"
FIGURES_DIR = ANALYSIS_ROOT / "figures"

DATA_DIR.mkdir(exist_ok=True)
FIGURES_DIR.mkdir(exist_ok=True)

print("Directory structure:")
print(f"  Analysis root: {ANALYSIS_ROOT.resolve()}")
print(f"  Data directory: {DATA_DIR.resolve()}")
print(f"  Figures directory: {FIGURES_DIR.resolve()}")
```

```python
# Test plot generation
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(8, 4))
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Test Plot - Environment Working!")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print("✓ Matplotlib working correctly")
```
